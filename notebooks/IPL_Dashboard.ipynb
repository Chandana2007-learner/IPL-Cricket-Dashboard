```python
import matplotlib.pyplot as plt

# Top 10 batsmen by runs
df.groupby('batter')['runs_batter'].sum().sort_values(ascending=False).head(10).plot(kind='bar', title='Top 10 Batsmen by runs')
plt.xlabel('batter')
plt.ylabel('runs')
plt.show()

# Team wwns
df['Winner'].value_counts().plot(kind='bar', title='Matches Won by Each Team')
plt.xlabel('team')
plt.ylabel('wins')
plt.show()

# Toss decision
df['Decision'].value_counts().plot(kind='bar', title='Bat vs Field decision')
plt.xlabel('decision')
plt.ylabel('count')
plt.show()

# Season-wise records
df['Season'].value_counts().sort_index().plot(kind='line', marker='o', title='IPL Data by season')
plt.xlabel('season')
plt.ylabel('records')
plt.show()
```
